```python
df = pd.read_csv("Bengaluru_House_Data.csv")
plt.hist(df["price"].dropna(), bins=30)
plt.title("Distribution of house prices")
plt.xlabel("Price (dataset units)")
plt.ylabel("Number of listings")
plt.show()
```

```python
avg_price = (
df.groupby("area_type")["price"]
.mean()
.sort_values()
)
avg_price.plot(kind="bar")
plt.title("Average price by area type")
plt.xlabel("Area type")
plt.ylabel("Average price")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

```python
plot_df = df[["total_sqft", "price"]].copy()
plot_df["total_sqft"] = pd.to_numeric(plot_df["total_sqft"], errors="coerce")
plot_df = plot_df.dropna()
plt.scatter(plot_df["total_sqft"], plot_df["price"],
alpha=0.4)
plt.title("House size vs. price")
plt.xlabel("Total area (sq ft)")
plt.ylabel("Price (dataset units)")
plt.show()
```
